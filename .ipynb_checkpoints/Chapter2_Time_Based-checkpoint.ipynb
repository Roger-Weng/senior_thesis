Preprocessing: Select weekdays 7AM-1:30PM

```python
import datetime as dt

def preprocessing(df):
    df_new = df.copy()
    df_new['datetime'] = df_new['date'] + df_new['time']
    df_new.drop(columns=['date', 'time'], inplace=True)

    df_new['datetime'] = pd.to_datetime(df_new['datetime'])

    mask = (
        ~df_new['datetime'].dt.date.isin([dt.date(2024, 2, 10),dt.date(2024, 2, 11)]) &
        (df_new['datetime'].dt.hour >= 7) & 
        ((df_new['datetime'].dt.hour < 13) | ((df_new['datetime'].dt.hour == 13) & (df_new['datetime'].dt.minute < 30)))
    )
    
    df_new = df_new[mask]
    
    return df_new

cl_df_ = preprocessing(cl_df)
print(cl_df_.isna().sum())
# ho_df_ = preprocessing(ho_df)
```

Quote snapshot features:
1. bid/ask price
2. bid/ask size
3. quote volume imbalance (top level)
4. micro price
5. mid price

```python
def quotes_snapshot_features(original_df, resampling = False):
    '''
    Snapshot features (quotes): downsampled to every 1 second
    bid
    bsiz
    ask
    asiz
    volume_imbalance: (bid volume - ask volume) / (bid volume + ask volume)
    micro_price: (ask price * bid volume + bid price * ask volume) / (bid volume + ask volume)
    mid_price: average price between bid and ask
    update_count: number of quote updates wihtin 1 second
    '''
    df = original_df.loc[original_df['prc'].isna(), ['datetime','bid', 'bsiz', 'ask', 'asiz']]
    den = df['bsiz'] + df['asiz']
    df['volume_imbalance'] = np.where(den>0, (df['bsiz']-df['asiz'])/den, np.nan)
    df['micro_price'] = np.where(den>0, (df['bsiz']*df['ask'] + df['asiz']*df['bid'])/den, np.nan)
    df['mid_price'] = (df['bid']+df['ask'])/2
    df['update_count'] = 1
    df = df.set_index('datetime')
    print(df)
    
    # with pd.option_context('display.max_rows', None):
    #     print(df.head(300))
    if resampling: 
        df = df.resample('1s', closed='right', label='right').agg({
                'bid': 'last',
                'bsiz': 'last',
                'ask': 'last',
                'asiz': 'last',
                'volume_imbalance': 'last',
                'micro_price': 'last',
                'mid_price': 'last',
                'update_count': 'sum'
            }).ffill()
        # print(df)
        df = df.between_time('7:00:00', "13:30:00")
        print(df)

    
    return df

cl_df_with_snapshot = quotes_snapshot_features(cl_df_)
print(cl_df_with_snapshot.isna().sum())

# ho_df_with_snapshot = quotes_snapshot_features(ho_df_)
```

Quote time-windowed features:
1. past return
2. past volatility
3. target 1-min return

```python
def time_to_second(time: str) -> float:
    if time.endswith('ms'):
        time_sec = float(time[:-2]) / 1000
    if time.endswith('s'):
        time_sec = float(time[:-1])
    elif time.endswith('min'):
        time_sec = 60 * float(time[:-3])
    elif time.endswith('h'):
        time_sec = 3600 * float(time[:-1])
    else:
        raise ValueError("Input unit not in sec, min or hour.")
    return time_sec

def quotes_time_windowed_features(original_df, 
                                  past_windows=['1s','4s','15s','1min','4min','15min'], 
                                  target_window='3s'):

    df = original_df.between_time('7:00', '13:30')

    # Log return (mid and micro)
    log_micro_price = np.log(df['micro_price'])
    log_return_micro_price = log_micro_price.diff()

    seconds_per_day = 24 * 3600

    # Past-window features
    for window in past_windows:
        # rolling window is automatically time-based if window is like '3s'
        df[f'log_return_{window}'] = log_return_micro_price.rolling(window).sum()

        time_sec = pd.to_timedelta(window).total_seconds()
        df[f'realized_volatility_{window}'] = np.sqrt(
            log_return_micro_price.pow(2).rolling(window).sum() * seconds_per_day / time_sec
        )

    # Target window (FUTURE return)
    delta = pd.to_timedelta(target_window)

    log_mid_price = np.log(df['mid_price'])

    df[f'target_{target_window}_log_return'] = (
        log_mid_price.shift(freq=-delta) - log_mid_price
    )

    return df

quotes_df = quotes_time_windowed_features(cl_df_with_snapshot)
print(quotes_df.loc[quotes_df.index.date == dt.date(2024,2,5)].head(5))
cl_df_with_snapshot.head(5)
```

Trade Features (in past window):
1. Number of buys/sells
2. Volume of buys/sells
3. VWAP of buys/sells
4. Last buy/sell price

```python
import datetime as dt
import numpy as np
import pandas as pd

def trades_features(
    original_df: pd.DataFrame,
    past_windows: list[str] = ['1s', '4s', '15s', '1min', '4min', '15min']
) -> pd.DataFrame:
    # Keep only actual trades (prc not NaN)
    df = original_df.loc[original_df['prc'].notna(), ['datetime', 'aggr', 'siz', 'prc']].copy()

    # Remove 2/10 and 2/11
    exclude = {dt.date(2024, 2, 10), dt.date(2024, 2, 11)}
    df = df[~df['datetime'].dt.date.isin(exclude)]

    # Use datetime as index, sort, then restrict to trading hours
    df = df.set_index('datetime').sort_index()
    df = df.between_time('07:00', '13:30')

    # Precompute buy/sell masks
    is_buy = (df['aggr'] == 'B')
    is_sell = (df['aggr'] == 'S')

    # Volumes per trade
    buy_vol = df['siz'].where(is_buy, 0.0)
    sell_vol = df['siz'].where(is_sell, 0.0)

    # Trade counts (1 per trade)
    buy_trades = is_buy.astype(int)
    sell_trades = is_sell.astype(int)

    # Last traded price on each side (for VWAP fallback)
    last_buy_price = df['prc'].where(is_buy).ffill()
    last_sell_price = df['prc'].where(is_sell).ffill()

    # Output frame indexed by trade timestamps
    result = pd.DataFrame(index=df.index)

    for tf in past_windows:
        window = tf  # time-based window, e.g. '4s', '1min', ...

        # Counts over past window
        result[f'buy_count_{tf}'] = buy_trades.rolling(window).sum()
        result[f'sell_count_{tf}'] = sell_trades.rolling(window).sum()

        # Volumes over past window
        result[f'buy_volumes_{tf}'] = buy_vol.rolling(window).sum()
        result[f'sell_volumes_{tf}'] = sell_vol.rolling(window).sum()

        # Volume imbalance over past window
        den = result[f'buy_volumes_{tf}'] + result[f'sell_volumes_{tf}']
        result[f'trade_volume_imbalance_{tf}'] = np.where(
            den > 0,
            (result[f'buy_volumes_{tf}'] - result[f'sell_volumes_{tf}']) / den,
            0.0
        )

        # VWAP over past window (buy side)
        buy_num = (df['prc'] * buy_vol).rolling(window).sum()
        buy_den = buy_vol.rolling(window).sum()
        result[f'buy_vwap_{tf}'] = np.where(
            buy_den > 0,
            buy_num / buy_den,
            last_buy_price
        )

        # VWAP over past window (sell side)
        sell_num = (df['prc'] * sell_vol).rolling(window).sum()
        sell_den = sell_vol.rolling(window).sum()
        result[f'sell_vwap_{tf}'] = np.where(
            sell_den > 0,
            sell_num / sell_den,
            last_sell_price
        )

    return result

cl_trades_df = trades_features(cl_df_)
cl_trades_df
```

### Feature engineering pipeline

```python
def feature_engineering(df, name, past_windows=['1s','4s','15s','1min', '4min','15min'], target_window='3s'):
    df = preprocessing(df)
    # print(f"{name}\n")
    # print(df.head())
    # print(df.info())
    
    df_with_snapshot = quotes_snapshot_features(df)
    # print(f"Snapshot data")
    # print(df_with_snapshot.head(5))
    # print(df_with_snapshot.isna().sum())
    
    quotes_df = quotes_time_windowed_features(df_with_snapshot, past_windows, target_window)
    trades_df = trades_features(df, past_windows)
    
    features_df = quotes_df.join(trades_df, how='outer')

    # print(features_df.head(50))
    # features_df = features_df.dropna()
    return features_df

sr_df_feature_engineered = feature_engineering(sr_df, name = "10 YEAR US INT RATE SWAP", past_windows=['10ms','30ms','100ms','300ms', '1s','3s'])
cl_df_feature_engineered = feature_engineering(cl_df, name = "Crude Oil Futures", past_windows=['1s','4s','15s','1min', '4min','15min'])

cl_df_feature_engineered 
```

## Correlation Analysis

```python
import pandas as pd
import numpy as np

def correlation_analysis(features_df, target_col='target_3s_log_return', top_k=50):
    df = features_df.dropna(subset=[target_col])
    corrs = df.corrwith(df[target_col], method='spearman')
    corrs_sorted = corrs.reindex(corrs.abs().sort_values(ascending=False).index)

    print(f"=== Top {top_k} correlated features (Spearman, signed) ===")
    for feat, val in corrs_sorted.head(top_k).items():
        print(f"{feat:40s}  {val:+.5f}")

correlation_analysis(cl_df_feature_engineered)
# correlation_analysis(sr_df_feature_engineered)
```

## Modeling

```python
def lightGBM (features_df):
    import lightgbm as lgb
    import pandas as pd
    import numpy as np
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.metrics import r2_score
    from scipy.stats import spearmanr
    
    # === Prepare data ===
    target = 'target_1min_log_return'
    features = [c for c in features_df.columns if c != target]
    
    X = features_df[features]
    y = features_df[target]
    
    # Drop NaNs from target alignment
    mask = ~y.isna()
    X, y = X[mask], y[mask]
    
    # === Time-series cross-validation ===
    tscv = TimeSeriesSplit(n_splits=5)
    r2s, ics = [], []
    oof_pred = np.zeros(len(X))   # store out-of-fold predictions (no leakage)
    
    for fold, (train_idx, test_idx) in enumerate(tscv.split(X)):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
        model = lgb.LGBMRegressor(
            n_estimators=1000,          # more trees, smaller steps
            learning_rate=0.02,        # smaller learning rate for stability
            num_leaves=15,              # smaller leaves = smoother model
            max_depth=6,                # limit tree depth
            subsample=0.7,              # stochastic bagging
            colsample_bytree=0.7,       # feature bagging
            # reg_lambda=1e5,            # L2 regularization
            # reg_alpha=1.0,              # L1 regularization
            min_child_samples=200,      # larger → smoother splits
            # min_split_gain=1e-6,        # require some improvement
            random_state=42
        )
    
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        oof_pred[test_idx] = y_pred  # save only out-of-sample preds
    
        r2 = r2_score(y_test, y_pred)
        ic, _ = spearmanr(y_test, y_pred)
        r2s.append(r2)
        ics.append(ic)
        print(f"Fold {fold+1}: R²={r2:.4f}, IC={ic:.4f}")
    
    print(f"\nAverage R²={np.mean(r2s):.4f}, Average IC={np.mean(ics):.4f}")

    best_model = lgb.LGBMRegressor(
            n_estimators=1000,          # more trees, smaller steps
            learning_rate=0.02,        # smaller learning rate for stability
            num_leaves=15,              # smaller leaves = smoother model
            max_depth=6,                # limit tree depth
            subsample=0.7,              # stochastic bagging
            colsample_bytree=0.7,       # feature bagging
            # reg_lambda=1e5,            # L2 regularization
            # reg_alpha=1.0,              # L1 regularization
            min_child_samples=200,      # larger → smoother splits
            # min_split_gain=1e-6,        # require some improvement
            random_state=42
    )
    best_model.fit(X, y)
    
    return model, oof_pred
```

```python
def ridgeRegression(features_df):
    import pandas as pd
    import numpy as np
    from sklearn.linear_model import Ridge
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.preprocessing import StandardScaler
    from sklearn.metrics import r2_score
    from scipy.stats import spearmanr
    
    # === Prepare data ===
    target = 'target_1min_log_return'
    features = [c for c in features_df.columns if c != target]
    
    X = features_df[features]
    y = features_df[target]
    
    # Drop NaNs from target alignment
    mask = ~y.isna()
    X, y = X[mask], y[mask]
    
    # === Standardization (important for Ridge) ===
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    # === Time-series cross-validation ===
    tscv = TimeSeriesSplit(n_splits=5)
    alphas = [1e3, 5e3, 1e4, 5e4, 1e5, 1e6, 1e7]
    
    results = []
    best_ic = -np.inf
    best_alpha = None
    
    for alpha in alphas:
        r2s, ics = [], []
    
        for fold, (train_idx, test_idx) in enumerate(tscv.split(X_scaled)):
            X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
            y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
            model = Ridge(alpha=alpha, fit_intercept=True, random_state=42)
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
    
            r2 = r2_score(y_test, y_pred)
            ic, _ = spearmanr(y_test, y_pred)
            r2s.append(r2)
            ics.append(ic)
    
        mean_r2, mean_ic = np.mean(r2s), np.mean(ics)
        results.append((alpha, mean_r2, mean_ic))
        print(f"alpha={alpha:<6} | avg R²={mean_r2:>8.4f} | avg IC={mean_ic:>8.4f}")
    
        if mean_ic > best_ic:
            best_ic = mean_ic
            best_alpha = alpha
    
    print("\n=== Ridge tuning results ===")
    for alpha, mean_r2, mean_ic in results:
        print(f"alpha={alpha:<6} | avg R²={mean_r2:>8.4f} | avg IC={mean_ic:>8.4f}")
    print(f"\nBest alpha = {best_alpha} (avg IC={best_ic:.4f})")
    
    # === Refit with best alpha, collect OOF preds ===
    r2s, ics = [], []
    oof_pred = np.zeros(len(X))
    
    for fold, (train_idx, test_idx) in enumerate(tscv.split(X_scaled)):
        X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
        model = Ridge(alpha=best_alpha, fit_intercept=True, random_state=42)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        oof_pred[test_idx] = y_pred
    
        r2 = r2_score(y_test, y_pred)
        ic, _ = spearmanr(y_test, y_pred)
        r2s.append(r2)
        ics.append(ic)
        print(f"Fold {fold+1}: R²={r2:.4f}, IC={ic:.4f}")
    
    print(f"\nAverage R²={np.mean(r2s):.4f}, Average IC={np.mean(ics):.4f}")
    
    # # === Combine predictions ===
    # features_df = features_df.loc[mask].copy()
    # features_df['pred'] = oof_pred

    return model, oof_pred
```

```python
def trading_strategy(oof_pred, features_df, threshold_std=2):
    # === Combine results ===
    mask = ~features_df['target_1min_log_return'].isna()
    df = features_df.loc[mask].copy()
    df['pred'] = oof_pred
    
    # === Trading rule ===
    df['spread'] = df['ask'] - df['bid']
    df['rel_spread'] = df['spread'] / df['mid_price']

    # choose a threshold for strongest signals (e.g. top 5–10%)
    threshold = threshold_std * df['pred'].std()
    
    df['signal'] = np.where(df['pred'] > threshold,  1,
                    np.where(df['pred'] < -threshold, -1, 0))
    
    # Strategy return: predicted signal * future return - half-spread cost
    df['strategy_ret'] = (
        df['signal'] * df['target_1min_log_return']
        # - (df['rel_spread']/2) * (df['signal'] != 0)
    )
    
    # === Evaluation ===
    active = df['signal'] != 0
    hit_ratio = np.mean(
        np.sign(df.loc[active, 'signal'] * df.loc[active, 'target_1min_log_return']) > 0
    )

    ic = df[['pred', 'target_1min_log_return']].corr(method='spearman').iloc[0, 1]
    sharpe = df['strategy_ret'].mean() / df['strategy_ret'].std() * np.sqrt(252*24*60)
    
    print("\n=== Strategy Evaluation (Out-of-Sample) ===")
    print(f"Hit ratio: {hit_ratio:.3f}")
    print(f"Average Spearman IC: {ic:.4f}")
    print(f"Approx Sharpe: {sharpe:.2f}")
    print(f"Mean strategy return: {df['strategy_ret'].mean():.9f}")
    print(f"Std strategy return:  {df['strategy_ret'].std():.9f}")
```
