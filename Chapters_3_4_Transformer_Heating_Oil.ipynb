```python
import datetime as dt

LARGE_NEG = -9223372036854775808

# ============================================================
# 1) Preprocessing time & fill out 'N' trade type to 'B'/'S'
# ============================================================
def preprocessing_time(df: pd.DataFrame) -> pd.DataFrame:
    df_new = df.copy()
    df_new['datetime'] = df_new['date'] + df_new['time']
    df_new.drop(columns=['date', 'time'], inplace=True)

    df_new['datetime'] = pd.to_datetime(df_new['datetime'])

    mask = (
        ~df_new['datetime'].dt.date.isin([dt.date(2024, 2, 10),dt.date(2024, 2, 11)]) &
        (df_new['datetime'].dt.hour >= 7) & 
        ((df_new['datetime'].dt.hour < 13) | ((df_new['datetime'].dt.hour == 13) & (df_new['datetime'].dt.minute < 30)))
    )
    
    df_new = df_new[mask]

    is_N_trades = (df_new["aggr"] == "N")

    print(f"Number of unlabelled trades {is_N_trades.sum()}")

    # Your original heuristic: use prev quote (shift(1)) vs next record (shift(-1))
    same_dt_as_next = (df_new["datetime"] == df_new["datetime"].shift(-1))

    ask_size_decrease = (
        (df_new["asiz"].shift(1) > df_new["asiz"].shift(-1)) &
        ((df_new["asiz"].shift(1) - df_new["asiz"].shift(-1)) == df_new["siz"])
    )
    bid_size_decrease = (
        (df_new["bsiz"].shift(1) > df_new["bsiz"].shift(-1)) &
        ((df_new["bsiz"].shift(1) - df_new["bsiz"].shift(-1)) == df_new["siz"])
    )

    ask_price_up = (df_new["ask"].shift(1) < df_new["ask"].shift(-1))
    bid_price_down = (df_new["bid"].shift(1) > df_new["bid"].shift(-1))

    cond_B_trades = is_N_trades & same_dt_as_next & (ask_size_decrease | ask_price_up)
    cond_S_trades = is_N_trades & same_dt_as_next & (bid_size_decrease | bid_price_down)

    # If both fire (rare), keep as N (ambiguous)
    both = cond_B_trades & cond_S_trades
    df_new.loc[both, "aggr"] = "N"
    df_new.loc[cond_B_trades & ~both, "aggr"] = "B"
    df_new.loc[cond_S_trades & ~both, "aggr"] = "S"

    return df_new
```

```python
# ============================================================
# 2) PREPROCESSING & LABELING Market Data
#    - Adds Trade_B_PxChg / Trade_B_PxSame
#    - Adds Trade_S_PxChg / Trade_S_PxSame
# ============================================================
def preprocess_market_data(df: pd.DataFrame) -> pd.DataFrame:
    print(">>> Preprocessing Data...")

    # Keep ordering stable
    sort_cols = ["datetime"] + (["seq"] if "seq" in df.columns else [])
    df = df.sort_values(sort_cols, kind="mergesort").reset_index(drop=True)

    # --- Clean INT64_MIN for relevant columns ---
    cols_to_clean = ["bsiz", "bid", "siz", "ask", "asiz"]
    cleaned = {}
    for c in cols_to_clean:
        if c not in df.columns:
            raise ValueError(f"Missing required column '{c}'")
        cleaned[c] = df[c].where(df[c] != LARGE_NEG, np.nan)

    # --- Time delta in microseconds (vectorized) ---
    dt_us = df["datetime"].diff().dt.total_seconds().fillna(0.0).to_numpy() * 1e6
    log_time = np.log1p(dt_us)

    n = len(df)
    event = np.full(n, "No_Change", dtype=object)
    log_size = np.zeros(n, dtype=np.float64)

    if n > 0:
        event[0] = "Start"
        log_size[0] = 0.0

    # --- Build "state" quotes via ffill only on quote cols ---
    quote_cols = ["bid", "bsiz", "ask", "asiz"]
    q = pd.DataFrame({c: cleaned[c] for c in quote_cols}, index=df.index).ffill()

    bid = q["bid"].to_numpy()
    ask = q["ask"].to_numpy()
    bsiz = q["bsiz"].to_numpy()
    asiz = q["asiz"].to_numpy()

    # prev values (shift by 1)
    bid_prev  = np.roll(bid, 1);  bid_prev[0]  = np.nan
    ask_prev  = np.roll(ask, 1);  ask_prev[0]  = np.nan
    bsiz_prev = np.roll(bsiz, 1); bsiz_prev[0] = np.nan
    asiz_prev = np.roll(asiz, 1); asiz_prev[0] = np.nan

    # next values (shift by -1) for "immediately after trade"
    bid_next = np.roll(bid, -1); bid_next[-1] = np.nan
    ask_next = np.roll(ask, -1); ask_next[-1] = np.nan

    bsiz_delta = bsiz - bsiz_prev
    asiz_delta = asiz - asiz_prev

    # --- Trade mask (trade rows have non-NaN siz) ---
    siz = cleaned["siz"].to_numpy()
    trade_mask = ~np.isnan(siz)

    # Side array (uppercase strings)
    side = df["aggr"].fillna("N").astype(str).str.upper().to_numpy(dtype=str)

    valid = (np.arange(n) > 0)

    # --- Quote events (same as your original) ---
    conds = [
        valid & (bid > bid_prev),
        valid & (bid < bid_prev),
        valid & (bsiz_delta > 0),
        valid & (bsiz_delta < 0),
        valid & (ask > ask_prev),
        valid & (ask < ask_prev),
        valid & (asiz_delta > 0),
        valid & (asiz_delta < 0),
    ]
    choices = [
        "Bid_Px_Up",
        "Bid_Px_Dn",
        "Bid_Sz_Up",
        "Bid_Sz_Dn",
        "Ask_Px_Up",
        "Ask_Px_Dn",
        "Ask_Sz_Up",
        "Ask_Sz_Dn",
    ]

    quote_event = np.select(conds, choices, default="No_Change")

    quote_sz = np.select(
        conds,
        [
            1.0, 1.0,
            np.abs(bsiz_delta), np.abs(bsiz_delta),
            1.0, 1.0,
            np.abs(asiz_delta), np.abs(asiz_delta),
        ],
        default=0.0
    )
    quote_log_size = np.log1p(quote_sz)

    event[:] = quote_event
    log_size[:] = quote_log_size

    # --- Override with enriched trade labels ---
    trade_apply = valid & trade_mask

    # Start from base labels
    base_trade = np.char.add("Trade_", side).astype(object)

    # Split B/S into PxChg vs PxSame using "immediately next record"
    is_B = trade_apply & (side == "B")
    is_S = trade_apply & (side == "S")

    # Your definition:
    #  - Sell: bid goes DOWN after sell
    #  - Buy:  ask goes UP after buy
    B_px_chg = is_B & (ask_next > ask_prev)
    S_px_chg = is_S & (bid_next < bid_prev)

    base_trade[B_px_chg] = "Trade_B_PxChg"
    base_trade[is_B & ~B_px_chg] = "Trade_B_PxSame"

    base_trade[S_px_chg] = "Trade_S_PxChg"
    base_trade[is_S & ~S_px_chg] = "Trade_S_PxSame"

    # (N or other labels remain Trade_N / Trade_*)
    event[trade_apply] = base_trade[trade_apply]
    log_size[trade_apply] = np.log1p(siz[trade_apply])

    # Ensure row 0 remains Start
    if n > 0:
        event[0] = "Start"
        log_size[0] = 0.0

    out = pd.DataFrame({
        "Event": event,
        "LogTime": log_time,
        "LogSize": log_size,
    })

    # Drop No_Change and Start
    out = out[~out["Event"].isin(["No_Change", "Start"])].reset_index(drop=True)
    return out
```

# Transformer

```python
import math
import os
import random
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

# =========================
# CONFIGURATION
# =========================
SEQ_LEN = 100
PRED_LEN = 1
BATCH_SIZE = 256

# Transformer hyperparameters
D_MODEL = 128
NHEAD = 4
NUM_LAYERS = 4
FFN_DIM = 256
DROPOUT = 0.10

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 1e-2
EPOCHS = 10
GRAD_CLIP = 1.0

# time-based split fractions
TRAIN_FRAC = 0.8
VAL_FRAC   = 0.1
TEST_FRAC  = 0.1

# Drop any sample window whose span contains ANY of these events
BANNED_EVENTS_IN_WINDOW = ["Trade_N", "Start", "No_Change"]

# Optional: also exclude these from the unconditional baseline distribution computation
BANNED_EVENTS_FOR_GLOBAL = ["Trade_N", "Start", "No_Change"]

warnings.filterwarnings("ignore")


# =========================
# 0. GPU / PERFORMANCE
# =========================
def seed_everything(seed: int = 123):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(123)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f">>> Using device: {DEVICE}")

if DEVICE.type == "cuda":
    torch.backends.cudnn.benchmark = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

USE_AMP = (DEVICE.type == "cuda")
scaler = torch.cuda.amp.GradScaler(enabled=USE_AMP)

NUM_WORKERS = min(8, os.cpu_count() or 1) if DEVICE.type == "cuda" else 0
PIN_MEMORY = (DEVICE.type == "cuda")


# =========================
# 1. DATASET
# =========================
class MarketDataset(Dataset):
    """
    Builds sliding-window samples:
      x: past SEQ_LEN events + continuous feats
      y: distribution of next PRED_LEN events
      rolling_prior: distribution of events inside x window

    Windows containing banned events can be dropped.
    """

    def __init__(
        self,
        df: pd.DataFrame,
        seq_len: int,
        pred_len: int,
        vocab=None,
        add_unk: bool = True,
        banned_events_in_window=None,
        banned_events_for_global=None,
    ):
        self.seq_len = int(seq_len)
        self.pred_len = int(pred_len)

        required = {"Event", "LogTime", "LogSize"}
        missing = required - set(df.columns)
        if missing:
            raise ValueError(f"Missing required columns in df: {missing}")

        events_str = df["Event"].astype(str)

        if vocab is None:
            uniq = pd.unique(events_str).tolist()
            self.vocab = {ev: i for i, ev in enumerate(uniq)}
            if add_unk and "<UNK>" not in self.vocab:
                self.vocab["<UNK>"] = len(self.vocab)
        else:
            self.vocab = dict(vocab)

        self.idx_to_event = {i: ev for ev, i in self.vocab.items()}
        self.num_classes = len(self.vocab)
        unk_id = self.vocab.get("<UNK>", None)

        mapped = events_str.map(self.vocab)
        if mapped.isna().any():
            if unk_id is None:
                unseen = events_str[mapped.isna()].unique()[:5]
                raise KeyError(f"Unseen events in this split but no <UNK> token. Examples: {unseen}")
            mapped = mapped.fillna(unk_id)

        event_ids_np = mapped.astype(np.int64).to_numpy()
        cont_np = df[["LogTime", "LogSize"]].to_numpy(dtype=np.float32)

        self.event_ids = torch.from_numpy(event_ids_np).long()
        self.cont_feats = torch.from_numpy(cont_np).float()

        L = len(self.event_ids)
        W = self.seq_len + self.pred_len
        if L < W:
            raise ValueError(f"Not enough rows: need at least seq_len+pred_len={W}, got {L}")

        banned_events_in_window = banned_events_in_window or []
        banned_ids = [self.vocab[e] for e in banned_events_in_window if e in self.vocab]

        if len(banned_ids) > 0:
            bad_mask = np.isin(event_ids_np, np.array(banned_ids, dtype=np.int64)).astype(np.int32)
            pref = np.empty(L + 1, dtype=np.int32)
            pref[0] = 0
            np.cumsum(bad_mask, out=pref[1:])
            window_sums = pref[W:] - pref[:-W]
            valid = (window_sums == 0)
            self.valid_starts = np.nonzero(valid)[0].astype(np.int64)
        else:
            self.valid_starts = np.arange(0, L - W + 1, dtype=np.int64)

        banned_events_for_global = banned_events_for_global or []
        banned_global_ids = [self.vocab[e] for e in banned_events_for_global if e in self.vocab]
        if len(banned_global_ids) > 0:
            good_mask = ~np.isin(event_ids_np, np.array(banned_global_ids, dtype=np.int64))
            base_ids = event_ids_np[good_mask]
        else:
            base_ids = event_ids_np

        counts = np.bincount(base_ids, minlength=self.num_classes).astype(np.float64)
        total = counts.sum()
        if total <= 0:
            self.global_dist = (np.ones(self.num_classes, dtype=np.float32) / self.num_classes)
        else:
            self.global_dist = (counts / total).astype(np.float32)

    def __len__(self):
        return int(len(self.valid_starts))

    def __getitem__(self, i: int):
        idx = int(self.valid_starts[i])

        x_events = self.event_ids[idx : idx + self.seq_len]
        x_cont   = self.cont_feats[idx : idx + self.seq_len]

        future_events = self.event_ids[idx + self.seq_len : idx + self.seq_len + self.pred_len]
        dist_counts = torch.bincount(future_events, minlength=self.num_classes).to(torch.float32)
        dist_vector = dist_counts / dist_counts.sum().clamp_min(1.0)

        prior_counts = torch.bincount(x_events, minlength=self.num_classes).to(torch.float32)
        rolling_prior = prior_counts / prior_counts.sum().clamp_min(1.0)

        return x_events, x_cont, dist_vector, rolling_prior


# =========================
# 2. MODEL: Decoder-only Transformer
# =========================
class SinusoidalPositionalEncoding(nn.Module):
    def __init__(self, d_model: int, max_len: int):
        super().__init__()
        pe = torch.zeros(max_len, d_model, dtype=torch.float32)
        position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div_term = torch.exp(
            torch.arange(0, d_model, 2, dtype=torch.float32) * (-math.log(10000.0) / d_model)
        )
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term[: pe[:, 1::2].shape[1]])
        self.register_buffer("pe", pe.unsqueeze(0), persistent=False)  # [1, L, D]

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return x + self.pe[:, : x.size(1)]


class MarketDecoderOnlyTransformer(nn.Module):
    """
    GPT-style causal self-attention over the input window only.
    It is "decoder-only" in the sense that it uses masked self-attention
    with no encoder and no cross-attention.

    Input:
      events:     [B, Seq]
      continuous: [B, Seq, 2]

    Output:
      log_probs over next-event class: [B, C]
    """
    def __init__(
        self,
        num_classes: int,
        d_model: int,
        nhead: int,
        num_layers: int,
        ffn_dim: int,
        dropout: float,
        max_seq_len: int,
    ):
        super().__init__()
        self.num_classes = num_classes
        self.d_model = d_model

        self.event_embedding = nn.Embedding(num_classes, d_model)
        self.continuous_projection = nn.Sequential(
            nn.Linear(2, d_model),
            nn.GELU(),
            nn.Linear(d_model, d_model),
        )
        self.input_norm = nn.LayerNorm(d_model)
        self.pos_encoding = SinusoidalPositionalEncoding(d_model, max_seq_len)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=ffn_dim,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=True,
        )
        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers,
            norm=nn.LayerNorm(d_model),
        )

        self.dropout = nn.Dropout(dropout)
        self.head = nn.Linear(d_model, num_classes)
        self.log_softmax = nn.LogSoftmax(dim=-1)

        self.register_buffer(
            "causal_mask",
            torch.triu(torch.full((max_seq_len, max_seq_len), float("-inf")), diagonal=1),
            persistent=False,
        )

    def forward(self, events: torch.Tensor, continuous: torch.Tensor) -> torch.Tensor:
        seq_len = events.size(1)
        tok = self.event_embedding(events)                 # [B, S, D]
        side = self.continuous_projection(continuous)      # [B, S, D]

        x = tok + side
        x = self.input_norm(x)
        x = self.pos_encoding(x)
        x = self.dropout(x)

        causal_mask = self.causal_mask[:seq_len, :seq_len]
        h = self.transformer(x, mask=causal_mask)          # [B, S, D]

        last_hidden = h[:, -1, :]                          # [B, D]
        logits = self.head(last_hidden)                    # [B, C]
        return self.log_softmax(logits)                    # [B, C] log-probs


# =========================
# 3. EVALUATION
# =========================
@torch.no_grad()
def evaluate(model, loader, criterion, global_log_prior):
    model.eval()
    sum_model = 0.0
    sum_roll  = 0.0
    sum_glob  = 0.0
    n = 0

    for x_ev, x_cont, y_target, y_rolling_prior in loader:
        bs = x_ev.size(0)
        x_ev = x_ev.to(DEVICE, non_blocking=True)
        x_cont = x_cont.to(DEVICE, non_blocking=True)
        y_target = y_target.to(DEVICE, non_blocking=True)
        y_rolling_prior = y_rolling_prior.to(DEVICE, non_blocking=True)

        log_probs = model(x_ev, x_cont)

        loss_model = criterion(log_probs, y_target)
        roll_log = torch.log(y_rolling_prior + 1e-9)
        loss_roll = criterion(roll_log, y_target)

        glob_log = global_log_prior.expand_as(log_probs)
        loss_glob = criterion(glob_log, y_target)

        sum_model += float(loss_model.item()) * bs
        sum_roll  += float(loss_roll.item())  * bs
        sum_glob  += float(loss_glob.item())  * bs
        n += bs

    return sum_model / n, sum_roll / n, sum_glob / n


# =========================
# 4. TRAINING
# =========================
def run_training(processed_df: pd.DataFrame):
    assert all(c in processed_df.columns for c in ["Event", "LogTime", "LogSize"]), \
        "processed_df must contain columns: Event, LogTime, LogSize"

    processed_df = processed_df.reset_index(drop=True)

    N = len(processed_df)
    train_end = int(TRAIN_FRAC * N)
    val_end   = int((TRAIN_FRAC + VAL_FRAC) * N)

    train_df = processed_df.iloc[:train_end].reset_index(drop=True)
    val_df   = processed_df.iloc[train_end:val_end].reset_index(drop=True)
    test_df  = processed_df.iloc[val_end:].reset_index(drop=True)

    print(f">>> Split sizes (rows): train={len(train_df)}, val={len(val_df)}, test={len(test_df)}")

    train_dataset = MarketDataset(
        train_df, SEQ_LEN, PRED_LEN,
        vocab=None,
        add_unk=True,
        banned_events_in_window=BANNED_EVENTS_IN_WINDOW,
        banned_events_for_global=BANNED_EVENTS_FOR_GLOBAL,
    )
    vocab = train_dataset.vocab

    val_dataset = MarketDataset(
        val_df, SEQ_LEN, PRED_LEN,
        vocab=vocab,
        add_unk=True,
        banned_events_in_window=BANNED_EVENTS_IN_WINDOW,
        banned_events_for_global=BANNED_EVENTS_FOR_GLOBAL,
    )
    test_dataset = MarketDataset(
        test_df, SEQ_LEN, PRED_LEN,
        vocab=vocab,
        add_unk=True,
        banned_events_in_window=BANNED_EVENTS_IN_WINDOW,
        banned_events_for_global=BANNED_EVENTS_FOR_GLOBAL,
    )

    print(f">>> Samples after dropping banned windows:")
    print(f"    train samples={len(train_dataset)}, val samples={len(val_dataset)}, test samples={len(test_dataset)}")

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )

    model = MarketDecoderOnlyTransformer(
        num_classes=train_dataset.num_classes,
        d_model=D_MODEL,
        nhead=NHEAD,
        num_layers=NUM_LAYERS,
        ffn_dim=FFN_DIM,
        dropout=DROPOUT,
        max_seq_len=SEQ_LEN,
    ).to(DEVICE)

    try:
        if hasattr(torch, "compile"):
            model = torch.compile(model)
    except Exception as e:
        print(f">>> torch.compile unavailable or failed (continuing without): {e}")

    optimizer = optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
        betas=(0.9, 0.95),
    )

    criterion = nn.KLDivLoss(reduction="batchmean")

    global_prior = torch.tensor(train_dataset.global_dist, dtype=torch.float32, device=DEVICE)
    global_log_prior = torch.log(global_prior + 1e-9).unsqueeze(0)

    best_val = float("inf")
    best_state = None

    print(f"\n>>> Starting Training (classes={train_dataset.num_classes})")
    print(f"Predict next {PRED_LEN} event distribution from past {SEQ_LEN}.")
    print(f"Model: decoder-only transformer with causal self-attention.")
    print(f"Banned events in any sample window: {BANNED_EVENTS_IN_WINDOW}\n")

    for epoch in range(EPOCHS):
        model.train()
        sum_train = 0.0
        n_train = 0

        for x_ev, x_cont, y_target, _y_roll in train_loader:
            bs = x_ev.size(0)

            x_ev = x_ev.to(DEVICE, non_blocking=True)
            x_cont = x_cont.to(DEVICE, non_blocking=True)
            y_target = y_target.to(DEVICE, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)

            with torch.cuda.amp.autocast(enabled=USE_AMP):
                log_probs = model(x_ev, x_cont)
                loss = criterion(log_probs, y_target)

            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=GRAD_CLIP)
            scaler.step(optimizer)
            scaler.update()

            sum_train += float(loss.item()) * bs
            n_train += bs

        train_kl = sum_train / max(1, n_train)
        val_model, val_roll, val_glob = evaluate(model, val_loader, criterion, global_log_prior)

        gain_vs_roll = val_roll - val_model
        gain_vs_glob = val_glob - val_model

        print(
            f"Epoch {epoch+1:02d}/{EPOCHS} | "
            f"Train KL: {train_kl:.4f} | "
            f"Val KL (Model/Roll/Global): {val_model:.4f} / {val_roll:.4f} / {val_glob:.4f} | "
            f"Gain (vs Roll/Global): {gain_vs_roll:.4f} / {gain_vs_glob:.4f}"
        )

        if val_model < best_val:
            best_val = val_model
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}

    if best_state is not None:
        model.load_state_dict(best_state)

    test_model, test_roll, test_glob = evaluate(model, test_loader, criterion, global_log_prior)
    print("\n>>> TEST RESULTS")
    print(f"TEST KL (Model/Roll/Global): {test_model:.4f} / {test_roll:.4f} / {test_glob:.4f}")
    print(f"TEST Gain (vs Roll/Global): {test_roll - test_model:.4f} / {test_glob - test_model:.4f}")

    return model, test_loader, test_dataset


# =========================
# 5. HELPER: build test loader/dataset only
# =========================
def build_test_loader_and_dataset(processed_df: pd.DataFrame):
    assert all(c in processed_df.columns for c in ["Event", "LogTime", "LogSize"]), \
        "processed_df must contain columns: Event, LogTime, LogSize"

    processed_df = processed_df.reset_index(drop=True)

    N = len(processed_df)
    train_end = int(TRAIN_FRAC * N)
    val_end   = int((TRAIN_FRAC + VAL_FRAC) * N)

    train_df = processed_df.iloc[:train_end].reset_index(drop=True)
    val_df   = processed_df.iloc[train_end:val_end].reset_index(drop=True)
    test_df  = processed_df.iloc[val_end:].reset_index(drop=True)

    train_dataset = MarketDataset(
        train_df, SEQ_LEN, PRED_LEN,
        vocab=None,
        add_unk=True,
        banned_events_in_window=BANNED_EVENTS_IN_WINDOW,
        banned_events_for_global=BANNED_EVENTS_FOR_GLOBAL,
    )
    vocab = train_dataset.vocab

    _val_dataset = MarketDataset(
        val_df, SEQ_LEN, PRED_LEN,
        vocab=vocab,
        add_unk=True,
        banned_events_in_window=BANNED_EVENTS_IN_WINDOW,
        banned_events_for_global=BANNED_EVENTS_FOR_GLOBAL,
    )
    test_dataset = MarketDataset(
        test_df, SEQ_LEN, PRED_LEN,
        vocab=vocab,
        add_unk=True,
        banned_events_in_window=BANNED_EVENTS_IN_WINDOW,
        banned_events_for_global=BANNED_EVENTS_FOR_GLOBAL,
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )
    return test_loader, test_dataset


# =========================
# 6. COLLECT PREDICTIONS
# =========================
@torch.no_grad()
def collect_all_predictions(model, loader, dataset):
    model.eval()

    all_true = []
    all_pred = []
    all_prob_true = []
    all_prob_pred = []

    for x_ev, x_cont, y_target, _ in loader:
        x_ev = x_ev.to(DEVICE, non_blocking=True)
        x_cont = x_cont.to(DEVICE, non_blocking=True)
        y_target = y_target.to(DEVICE, non_blocking=True)

        log_probs = model(x_ev, x_cont)
        probs = torch.exp(log_probs)

        pred_idx = torch.argmax(probs, dim=1)
        true_idx = torch.argmax(y_target, dim=1)

        for i in range(x_ev.size(0)):
            true_event = dataset.idx_to_event[int(true_idx[i])]
            pred_event = dataset.idx_to_event[int(pred_idx[i])]
            prob_true  = probs[i, true_idx[i]].item()
            prob_pred  = probs[i, pred_idx[i]].item()

            all_true.append(true_event)
            all_pred.append(pred_event)
            all_prob_true.append(prob_true)
            all_prob_pred.append(prob_pred)

    df = pd.DataFrame({
        "True": all_true,
        "Predicted": all_pred,
        "Prob_True": all_prob_true,
        "Prob_Pred": all_prob_pred
    })
    return df


# =========================
# 7. OPTIONAL: autoregressive generation helper
# =========================
@torch.no_grad()
def generate_next_events(
    model,
    dataset,
    seed_events: torch.Tensor,
    seed_cont: torch.Tensor,
    steps: int = 10,
    temperature: float = 1.0,
    top_k: int = None,
    sample: bool = False,
):
    """
    Exploratory autoregressive rollout from a seed window.
    Because future LogTime/LogSize are unknown at generation time, generated steps
    are fed back with zero continuous features.

    seed_events: [Seq] long
    seed_cont:   [Seq, 2] float

    Returns a list of generated event strings.
    """
    model.eval()

    x_ev = seed_events.clone().long().to(DEVICE).unsqueeze(0)
    x_ct = seed_cont.clone().float().to(DEVICE).unsqueeze(0)

    out_events = []

    for _ in range(steps):
        log_probs = model(x_ev[:, -SEQ_LEN:], x_ct[:, -SEQ_LEN:])
        logits = log_probs / max(temperature, 1e-6)
        probs = torch.softmax(logits, dim=-1)

        if top_k is not None and top_k < probs.size(-1):
            top_vals, top_idx = torch.topk(probs, k=top_k, dim=-1)
            top_vals = top_vals / top_vals.sum(dim=-1, keepdim=True)
            if sample:
                choice = torch.multinomial(top_vals, num_samples=1)
                next_id = top_idx.gather(1, choice)
            else:
                next_id = top_idx[:, :1]
        else:
            if sample:
                next_id = torch.multinomial(probs, num_samples=1)
            else:
                next_id = torch.argmax(probs, dim=-1, keepdim=True)

        next_cont = torch.zeros((1, 1, 2), dtype=x_ct.dtype, device=DEVICE)

        x_ev = torch.cat([x_ev, next_id], dim=1)
        x_ct = torch.cat([x_ct, next_cont], dim=1)

        out_events.append(dataset.idx_to_event[int(next_id.item())])

    return out_events


# =========================
# 8. RUN
# =========================
if __name__ == "__main__":
    assert "processed_df" in globals(), "Please define processed_df before running this script."
    trained_model, test_loader, test_dataset = run_training(processed_df)
    pred_df = collect_all_predictions(trained_model, test_loader, test_dataset)
```

```python
# ============================================================
# Trade-direction evaluation vs "last-trade sign" baseline
#   - Evaluates ONLY when the target next event is a trade (B/S)
#   - Collapses model probs over Trade_B* and Trade_S* into p(B), p(S)
#   - Baseline: predict next trade sign = last trade sign seen in x window
#     (fallback to unconditional trade-direction prior if no prior trade in window)
# ============================================================

import math
import numpy as np
import pandas as pd
import torch


def _get_trade_sign_id_sets(dataset):
    """
    Buy-trade classes: any event string starting with 'Trade_B'
    Sell-trade classes: any event string starting with 'Trade_S'
    Works for: Trade_B, Trade_S, Trade_B_PxChg, Trade_B_PxSame, Trade_Sell, etc.
    """
    buy_ids, sell_ids = [], []
    for ev, idx in dataset.vocab.items():
        if isinstance(ev, str):
            if ev.startswith("Trade_B"):
                buy_ids.append(idx)
            elif ev.startswith("Trade_S"):
                sell_ids.append(idx)
    return buy_ids, sell_ids


def _trade_direction_prior_from_global_dist(dataset, buy_ids, sell_ids, eps=1e-12):
    """
    Computes P(B|trade), P(S|trade) from dataset.global_dist if available,
    otherwise falls back to 0.5/0.5.
    """
    if hasattr(dataset, "global_dist") and dataset.global_dist is not None:
        g = np.asarray(dataset.global_dist, dtype=np.float64)
        pB = float(g[buy_ids].sum()) if len(buy_ids) else 0.0
        pS = float(g[sell_ids].sum()) if len(sell_ids) else 0.0
        denom = pB + pS
        if denom > eps:
            return pB / denom, pS / denom
    return 0.5, 0.5


@torch.no_grad()
def evaluate_trade_sign_vs_last_trade(
    model,
    loader,
    dataset,
    eps: float = 1e-9,
    return_df: bool = True,
):
    """
    Returns:
      summary dict + (optional) per-sample DataFrame.

    Metrics computed on TRADE-ONLY subset:
      - model_logloss: binary log loss on trade direction (B vs S)
      - base_logloss:  binary log loss for baseline (copy last trade sign)
      - info_gain:     base_logloss - model_logloss  (positive is good)
      - model_acc / base_acc on trade direction
      - n_trade_targets: number of evaluated samples
      - frac_no_prior_trade: fraction where baseline had to use unconditional prior
    """
    model.eval()

    buy_ids, sell_ids = _get_trade_sign_id_sets(dataset)
    if len(buy_ids) == 0 or len(sell_ids) == 0:
        raise ValueError(
            f"Could not find both buy and sell trade classes in vocab. "
            f"Found |buy_ids|={len(buy_ids)}, |sell_ids|={len(sell_ids)}. "
            f"Make sure your events are named like 'Trade_B*' and 'Trade_S*'."
        )

    buy_ids_t = torch.tensor(buy_ids, dtype=torch.long, device=DEVICE)
    sell_ids_t = torch.tensor(sell_ids, dtype=torch.long, device=DEVICE)

    # Baseline fallback: unconditional P(B|trade), P(S|trade)
    pB_prior, pS_prior = _trade_direction_prior_from_global_dist(dataset, buy_ids, sell_ids)
    pB_prior = float(np.clip(pB_prior, eps, 1.0 - eps))
    pS_prior = float(np.clip(pS_prior, eps, 1.0 - eps))

    sum_model_ll = 0.0
    sum_base_ll = 0.0
    sum_model_acc = 0
    sum_base_acc = 0
    n_eval = 0
    n_no_prior = 0

    rows = [] if return_df else None

    for x_ev, x_cont, y_target, _ in loader:
        x_ev = x_ev.to(DEVICE, non_blocking=True)         # [B, Seq]
        x_cont = x_cont.to(DEVICE, non_blocking=True)     # [B, Seq, d]
        y_target = y_target.to(DEVICE, non_blocking=True) # [B, C] (one-hot if PRED_LEN=1)

        log_probs = model(x_ev, x_cont)                   # [B, C]
        probs = torch.exp(log_probs)                      # [B, C]

        # Collapse to trade-direction probs for model: p(B|trade), p(S|trade)
        pB_model_raw = probs.index_select(1, buy_ids_t).sum(dim=1)    # [B]
        pS_model_raw = probs.index_select(1, sell_ids_t).sum(dim=1)   # [B]
        pTrade = (pB_model_raw + pS_model_raw).clamp_min(eps)
        pB_model = (pB_model_raw / pTrade).clamp(eps, 1.0 - eps)      # [B]

        # Determine if target is a trade (buy or sell), using y_target distribution
        yB = y_target.index_select(1, buy_ids_t).sum(dim=1)           # [B]
        yS = y_target.index_select(1, sell_ids_t).sum(dim=1)          # [B]
        is_trade_target = (yB + yS) > 0.0                             # [B]

        if not is_trade_target.any():
            continue

        # Convert to hard label for accuracy (PRED_LEN=1 => one-hot)
        # true_buy = 1 if buy trade, else 0 (sell trade)
        true_buy = (yB > yS).to(torch.float32)                        # [B], only meaningful where is_trade_target

        # Baseline p(B): copy last trade sign within x window; fallback to unconditional prior if none.
        x_ev_cpu = x_ev.detach().cpu().numpy()                        # [B, Seq]
        Bset = set(buy_ids)
        Sset = set(sell_ids)

        pB_base_list = []
        last_sign_list = []

        for seq in x_ev_cpu:
            last = None
            # scan backwards to find last trade event in history
            for eid in seq[::-1]:
                if int(eid) in Bset:
                    last = "B"
                    break
                if int(eid) in Sset:
                    last = "S"
                    break

            if last is None:
                n_no_prior += 1
                pB_base_list.append(pB_prior)
                last_sign_list.append("None")
            elif last == "B":
                pB_base_list.append(1.0 - eps)
                last_sign_list.append("B")
            else:
                pB_base_list.append(eps)
                last_sign_list.append("S")

        pB_base = torch.tensor(pB_base_list, dtype=torch.float32, device=DEVICE).clamp(eps, 1.0 - eps)

        # Now score ONLY trade-target samples in this batch
        idxs = torch.nonzero(is_trade_target, as_tuple=False).squeeze(1)

        tb = true_buy.index_select(0, idxs)            # [K]
        pm = pB_model.index_select(0, idxs)            # [K]
        pb = pB_base.index_select(0, idxs)             # [K]

        # binary log loss (cross-entropy) with hard labels
        # loss = -[ y log p + (1-y) log (1-p) ]
        model_ll = -(tb * torch.log(pm) + (1.0 - tb) * torch.log(1.0 - pm))
        base_ll  = -(tb * torch.log(pb) + (1.0 - tb) * torch.log(1.0 - pb))

        sum_model_ll += float(model_ll.sum().item())
        sum_base_ll  += float(base_ll.sum().item())

        # accuracy
        pred_model = (pm >= 0.5).to(torch.float32)
        pred_base  = (pb >= 0.5).to(torch.float32)

        sum_model_acc += int((pred_model == tb).sum().item())
        sum_base_acc  += int((pred_base == tb).sum().item())

        n_eval += int(tbsz := tb.numel())

        if return_df:
            # For interpretability, also attach true/pred event strings using argmax over full class space
            true_idx_full = torch.argmax(y_target, dim=1).detach().cpu().numpy()
            pred_idx_full = torch.argmax(probs, dim=1).detach().cpu().numpy()

            pm_cpu = pm.detach().cpu().numpy()
            pb_cpu = pb.detach().cpu().numpy()
            tb_cpu = tb.detach().cpu().numpy().astype(int)

            idxs_cpu = idxs.detach().cpu().numpy()
            for j, bi in enumerate(idxs_cpu):
                rows.append({
                    "True_Event": dataset.idx_to_event[int(true_idx_full[bi])],
                    "Pred_Event": dataset.idx_to_event[int(pred_idx_full[bi])],
                    "True_Sign": "B" if tb_cpu[j] == 1 else "S",
                    "Model_pB": float(pm_cpu[j]),
                    "Base_pB": float(pb_cpu[j]),
                    "Last_Trade_Sign_in_Window": last_sign_list[int(bi)],
                    "Model_LogLoss": float(model_ll[j].item()),
                    "Base_LogLoss": float(base_ll[j].item()),
                })


    if n_eval == 0:
        raise RuntimeError("No trade-target samples found in this loader (targets never Trade_B*/Trade_S*).")

    model_logloss = sum_model_ll / n_eval
    base_logloss  = sum_base_ll / n_eval
    info_gain     = base_logloss - model_logloss
    model_acc     = sum_model_acc / n_eval
    base_acc      = sum_base_acc / n_eval
    frac_no_prior = n_no_prior / max(1, n_eval)  # counted per-sample when building baseline, close enough

    summary = {
        "n_trade_targets": int(n_eval),
        "model_logloss": float(model_logloss),
        "baseline_logloss": float(base_logloss),
        "info_gain (baseline - model)": float(info_gain),
        "model_acc": float(model_acc),
        "baseline_acc": float(base_acc),
        "fallback_prior_pB": float(pB_prior),
        "frac_no_prior_trade_in_window": float(frac_no_prior),
    }

    out_df = pd.DataFrame(rows) if return_df else None
    return summary, out_df


# =========================
# USAGE (after you have trained_model, test_loader, test_dataset)
# =========================
trade_sign_summary, trade_sign_df = evaluate_trade_sign_vs_last_trade(
    trained_model, test_loader, test_dataset, return_df=True
)

print("\n>>> Trade-direction eval vs last-trade-sign baseline")
for k, v in trade_sign_summary.items():
    if isinstance(v, float):
        print(f"{k}: {v:.6f}")
    else:
        print(f"{k}: {v}")

# Optional: inspect hardest cases where model underperforms baseline
if trade_sign_df is not None and len(trade_sign_df) > 0:
    print("\n>>> Examples where baseline beats model most (by logloss)")
    print(
        trade_sign_df.assign(delta=lambda d: d["Model_LogLoss"] - d["Base_LogLoss"])
        .sort_values("delta", ascending=False)
        .head(10)
        .to_string(index=False)
    )
```

# Transformer (Predict Trade directions & Number of quote updates before trade)

```python
import math
import os
import random
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

# =========================
# CONFIGURATION
# =========================
SEQ_LEN = 100
BATCH_SIZE = 256

# Transformer hyperparameters
D_MODEL = 128
NHEAD = 4
NUM_LAYERS = 4
FFN_DIM = 256
DROPOUT = 0.10

LEARNING_RATE = 3e-4
WEIGHT_DECAY = 1e-2
EPOCHS = 10
GRAD_CLIP = 1.0

# Split fractions
TRAIN_FRAC = 0.8
VAL_FRAC   = 0.1
TEST_FRAC  = 0.1

# History windows containing these are dropped
HISTORY_BANNED_EVENTS = ["Trade_N"]

# Total loss = sign CE + GAP_LOSS_WEIGHT * quote-gap loss
GAP_LOSS_WEIGHT = 1.0

warnings.filterwarnings("ignore")


# =========================
# 0. GPU / PERFORMANCE
# =========================
def seed_everything(seed: int = 123):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(123)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f">>> Using device: {DEVICE}")

if DEVICE.type == "cuda":
    torch.backends.cudnn.benchmark = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

USE_AMP = (DEVICE.type == "cuda")
scaler = torch.cuda.amp.GradScaler(enabled=USE_AMP)

NUM_WORKERS = min(8, os.cpu_count() or 1) if DEVICE.type == "cuda" else 0
PIN_MEMORY = (DEVICE.type == "cuda")


# =========================
# 1. HELPERS
# =========================
def classify_event_string(ev: str) -> int:
    """
    Returns:
      0 = quote event
      1 = buy trade   (Trade_B*)
      2 = sell trade  (Trade_S*)
      3 = other trade / unlabeled trade (Trade_N or any other Trade_*)
    """
    if ev.startswith("Trade_B"):
        return 1
    if ev.startswith("Trade_S"):
        return 2
    if ev.startswith("Trade_"):
        return 3
    return 0


def get_buy_sell_trade_id_sets(vocab):
    buy_ids, sell_ids = set(), set()
    for ev, idx in vocab.items():
        if isinstance(ev, str):
            if ev.startswith("Trade_B"):
                buy_ids.add(int(idx))
            elif ev.startswith("Trade_S"):
                sell_ids.add(int(idx))
    return buy_ids, sell_ids


# =========================
# 2. DATASET
# =========================
class MarketTradeGapDataset(Dataset):
    """
    Each sample uses the past SEQ_LEN processed events as input.

    Target is defined by looking forward from the position immediately after the history:
      - sign of the FIRST labeled trade ahead (buy/sell)
      - number of quote events before that trade

    Examples:
      [history] | Trade_B_*                       -> sign=B, gap=0
      [history] | Quote, Quote, Quote, Trade_S_* -> sign=S, gap=3

    If the first future trade is Trade_N / unlabeled, the sample is dropped.
    """

    def __init__(
        self,
        df: pd.DataFrame,
        seq_len: int,
        vocab=None,
        add_unk: bool = True,
        banned_history_events=None,
    ):
        self.seq_len = int(seq_len)

        required = {"Event", "LogTime", "LogSize"}
        missing = required - set(df.columns)
        if missing:
            raise ValueError(f"Missing required columns in df: {missing}")

        events_str = df["Event"].astype(str).reset_index(drop=True)

        if vocab is None:
            uniq = pd.unique(events_str).tolist()
            self.vocab = {ev: i for i, ev in enumerate(uniq)}
            if add_unk and "<UNK>" not in self.vocab:
                self.vocab["<UNK>"] = len(self.vocab)
        else:
            self.vocab = dict(vocab)

        self.idx_to_event = {i: ev for ev, i in self.vocab.items()}
        self.num_event_tokens = len(self.vocab)
        unk_id = self.vocab.get("<UNK>", None)

        mapped = events_str.map(self.vocab)
        if mapped.isna().any():
            if unk_id is None:
                unseen = events_str[mapped.isna()].unique()[:5]
                raise KeyError(f"Unseen events in split but no <UNK> token. Examples: {unseen}")
            mapped = mapped.fillna(unk_id)

        event_ids_np = mapped.astype(np.int64).to_numpy()
        cont_np = df[["LogTime", "LogSize"]].to_numpy(dtype=np.float32)
        event_str_np = events_str.to_numpy(dtype=object)

        self.event_ids = torch.from_numpy(event_ids_np).long()
        self.cont_feats = torch.from_numpy(cont_np).float()

        n = len(event_ids_np)
        if n <= self.seq_len:
            raise ValueError(f"Need more than seq_len={self.seq_len} rows, got {n}")

        # 0 quote, 1 buy, 2 sell, 3 unlabeled/other trade
        event_type = np.fromiter(
            (classify_event_string(ev) for ev in event_str_np),
            dtype=np.int8,
            count=n,
        )

        # target_from_pos[t] means:
        # looking forward from row t (inclusive), what is the next labeled trade sign,
        # and how many quote events come before it?
        target_valid = np.zeros(n, dtype=bool)
        target_sign = np.full(n, -1, dtype=np.int64)  # 1=buy, 0=sell
        target_gap = np.zeros(n, dtype=np.int64)

        for t in range(n - 1, -1, -1):
            typ = event_type[t]

            if typ == 1:  # buy trade right here
                target_valid[t] = True
                target_sign[t] = 1
                target_gap[t] = 0

            elif typ == 2:  # sell trade right here
                target_valid[t] = True
                target_sign[t] = 0
                target_gap[t] = 0

            elif typ == 3:  # unlabeled / unusable trade blocks the sample
                target_valid[t] = False

            else:  # quote
                if t + 1 < n and target_valid[t + 1]:
                    target_valid[t] = True
                    target_sign[t] = target_sign[t + 1]
                    target_gap[t] = target_gap[t + 1] + 1

        self.target_valid_from_pos = target_valid
        self.target_sign_from_pos = target_sign
        self.target_gap_from_pos = target_gap

        # Ban certain events from the history window
        banned_history_events = banned_history_events or []
        banned_ids = [self.vocab[e] for e in banned_history_events if e in self.vocab]

        if len(banned_ids) > 0:
            bad_mask = np.isin(event_ids_np, np.array(banned_ids, dtype=np.int64)).astype(np.int32)
            pref = np.empty(n + 1, dtype=np.int32)
            pref[0] = 0
            np.cumsum(bad_mask, out=pref[1:])
            # count of banned events in each length-seq_len history block
            history_bad_count = pref[self.seq_len:] - pref[:-self.seq_len]  # length n-seq_len+1
            history_ok = (history_bad_count == 0)
        else:
            history_ok = np.ones(n - self.seq_len + 1, dtype=bool)

        # Sample start i uses history [i, i+seq_len-1], target search starts at i+seq_len
        start_positions = np.arange(0, n - self.seq_len, dtype=np.int64)
        target_positions = start_positions + self.seq_len

        valid_sample_mask = history_ok[:-1] & target_valid[target_positions]

        self.valid_starts = start_positions[valid_sample_mask]
        self.sample_target_positions = target_positions[valid_sample_mask]

        if len(self.valid_starts) == 0:
            raise ValueError("No valid samples after filtering.")

        self.sign_targets = self.target_sign_from_pos[self.sample_target_positions]
        self.gap_targets = self.target_gap_from_pos[self.sample_target_positions]

        sign_counts = np.bincount(self.sign_targets, minlength=2).astype(np.float64)
        total_sign = sign_counts.sum()
        self.sign_prior = (sign_counts / total_sign) if total_sign > 0 else np.array([0.5, 0.5], dtype=np.float64)

        self.gap_mean = float(np.mean(self.gap_targets)) if len(self.gap_targets) else 0.0
        self.gap_median = float(np.median(self.gap_targets)) if len(self.gap_targets) else 0.0

    def __len__(self):
        return int(len(self.valid_starts))

    def __getitem__(self, i: int):
        start = int(self.valid_starts[i])
        target_pos = int(self.sample_target_positions[i])

        x_events = self.event_ids[start : start + self.seq_len]
        x_cont   = self.cont_feats[start : start + self.seq_len]

        # sign target: 1=buy, 0=sell
        y_sign = torch.tensor(int(self.target_sign_from_pos[target_pos]), dtype=torch.long)

        # gap target: number of quote updates before next labeled trade
        y_gap = torch.tensor(float(self.target_gap_from_pos[target_pos]), dtype=torch.float32)

        return x_events, x_cont, y_sign, y_gap


# =========================
# 3. MODEL
# =========================
class SinusoidalPositionalEncoding(nn.Module):
    def __init__(self, d_model: int, max_len: int):
        super().__init__()
        pe = torch.zeros(max_len, d_model, dtype=torch.float32)
        position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div_term = torch.exp(
            torch.arange(0, d_model, 2, dtype=torch.float32) * (-math.log(10000.0) / d_model)
        )
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term[: pe[:, 1::2].shape[1]])
        self.register_buffer("pe", pe.unsqueeze(0), persistent=False)  # [1, L, D]

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return x + self.pe[:, : x.size(1)]


class MarketDecoderOnlyTransformerTradeGap(nn.Module):
    """
    Decoder-only transformer over the history window.
    Two output heads:
      1) sign_head: next trade sign (buy/sell)
      2) gap_head:  log(1 + number of quote updates until that trade)
    """

    def __init__(
        self,
        num_event_tokens: int,
        d_model: int,
        nhead: int,
        num_layers: int,
        ffn_dim: int,
        dropout: float,
        max_seq_len: int,
    ):
        super().__init__()
        self.event_embedding = nn.Embedding(num_event_tokens, d_model)

        self.continuous_projection = nn.Sequential(
            nn.Linear(2, d_model),
            nn.GELU(),
            nn.Linear(d_model, d_model),
        )

        self.input_norm = nn.LayerNorm(d_model)
        self.pos_encoding = SinusoidalPositionalEncoding(d_model, max_seq_len)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=ffn_dim,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=True,
        )
        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers,
            norm=nn.LayerNorm(d_model),
        )

        self.dropout = nn.Dropout(dropout)

        self.sign_head = nn.Linear(d_model, 2)  # [sell, buy] via target 0/1

        self.gap_head = nn.Sequential(
            nn.Linear(d_model, d_model),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_model, 1),
        )

        self.register_buffer(
            "causal_mask",
            torch.triu(torch.full((max_seq_len, max_seq_len), float("-inf")), diagonal=1),
            persistent=False,
        )

    def forward(self, events: torch.Tensor, continuous: torch.Tensor):
        seq_len = events.size(1)

        tok = self.event_embedding(events)            # [B, S, D]
        side = self.continuous_projection(continuous) # [B, S, D]

        x = tok + side
        x = self.input_norm(x)
        x = self.pos_encoding(x)
        x = self.dropout(x)

        h = self.transformer(x, mask=self.causal_mask[:seq_len, :seq_len])  # [B, S, D]
        last_hidden = h[:, -1, :]                                            # [B, D]

        sign_logits = self.sign_head(last_hidden)                             # [B, 2]
        gap_log_pred = self.gap_head(last_hidden).squeeze(-1)                # [B]

        return sign_logits, gap_log_pred


# =========================
# 4. EVALUATION HELPERS
# =========================
def get_last_trade_sign_baseline_probs(x_ev_np: np.ndarray, dataset: MarketTradeGapDataset):
    """
    Baseline for sign:
      predict next trade sign = last labeled trade sign seen in the history
      fallback = train-set unconditional P(buy)
    """
    buy_ids, sell_ids = get_buy_sell_trade_id_sets(dataset.vocab)

    prior_p_buy = float(dataset.sign_prior[1])
    p_buy = np.full(len(x_ev_np), prior_p_buy, dtype=np.float32)
    has_prior_trade = np.zeros(len(x_ev_np), dtype=bool)

    for i, seq in enumerate(x_ev_np):
        for eid in seq[::-1]:
            eid = int(eid)
            if eid in buy_ids:
                p_buy[i] = 1.0
                has_prior_trade[i] = True
                break
            if eid in sell_ids:
                p_buy[i] = 0.0
                has_prior_trade[i] = True
                break

    return p_buy, has_prior_trade


@torch.no_grad()
def evaluate_trade_gap_model(model, loader, dataset, return_df: bool = False):
    model.eval()

    ce_sum = 0.0
    gap_loss_sum = 0.0
    total_loss_sum = 0.0
    n = 0

    model_correct = 0
    base_correct = 0

    model_logloss_sum = 0.0
    base_logloss_sum = 0.0

    model_mae_sum = 0.0
    model_rmse_sum = 0.0

    base_mae_sum = 0.0
    base_rmse_sum = 0.0

    n_no_prior_trade = 0

    ce_loss_fn = nn.CrossEntropyLoss(reduction="sum")
    gap_loss_fn = nn.SmoothL1Loss(reduction="sum")

    rows = [] if return_df else None

    for x_ev, x_cont, y_sign, y_gap in loader:
        bs = x_ev.size(0)

        x_ev = x_ev.to(DEVICE, non_blocking=True)
        x_cont = x_cont.to(DEVICE, non_blocking=True)
        y_sign = y_sign.to(DEVICE, non_blocking=True)
        y_gap = y_gap.to(DEVICE, non_blocking=True)

        sign_logits, gap_log_pred = model(x_ev, x_cont)
        sign_probs = torch.softmax(sign_logits, dim=-1)

        true_gap_log = torch.log1p(y_gap)

        ce = ce_loss_fn(sign_logits, y_sign)
        gap_loss = gap_loss_fn(gap_log_pred, true_gap_log)
        total = ce + GAP_LOSS_WEIGHT * gap_loss

        ce_sum += float(ce.item())
        gap_loss_sum += float(gap_loss.item())
        total_loss_sum += float(total.item())

        # model sign metrics
        pred_sign = torch.argmax(sign_probs, dim=1)
        model_correct += int((pred_sign == y_sign).sum().item())

        p_buy_model = sign_probs[:, 1].clamp(1e-9, 1 - 1e-9)
        true_is_buy = y_sign.to(torch.float32)
        model_logloss = -(true_is_buy * torch.log(p_buy_model) + (1.0 - true_is_buy) * torch.log(1.0 - p_buy_model))
        model_logloss_sum += float(model_logloss.sum().item())

        # baseline sign metrics
        p_buy_base_np, has_prior_trade = get_last_trade_sign_baseline_probs(
            x_ev.detach().cpu().numpy(), dataset
        )
        n_no_prior_trade += int((~has_prior_trade).sum())

        p_buy_base = torch.tensor(p_buy_base_np, dtype=torch.float32, device=DEVICE).clamp(1e-9, 1 - 1e-9)
        pred_base = (p_buy_base >= 0.5).long()
        base_correct += int((pred_base == y_sign).sum().item())

        base_logloss = -(true_is_buy * torch.log(p_buy_base) + (1.0 - true_is_buy) * torch.log(1.0 - p_buy_base))
        base_logloss_sum += float(base_logloss.sum().item())

        # model gap metrics (evaluate on raw count scale)
        pred_gap = torch.expm1(gap_log_pred).clamp_min(0.0)
        abs_err = torch.abs(pred_gap - y_gap)
        sq_err = (pred_gap - y_gap) ** 2

        model_mae_sum += float(abs_err.sum().item())
        model_rmse_sum += float(sq_err.sum().item())

        # baseline gap: constant train-set median gap
        base_gap = torch.full_like(y_gap, fill_value=float(dataset.gap_median))
        base_abs_err = torch.abs(base_gap - y_gap)
        base_sq_err = (base_gap - y_gap) ** 2

        base_mae_sum += float(base_abs_err.sum().item())
        base_rmse_sum += float(base_sq_err.sum().item())

        if return_df:
            for i in range(bs):
                rows.append({
                    "True_Sign": "B" if int(y_sign[i].item()) == 1 else "S",
                    "Model_pBuy": float(p_buy_model[i].item()),
                    "Base_pBuy": float(p_buy_base[i].item()),
                    "Model_Sign": "B" if int(pred_sign[i].item()) == 1 else "S",
                    "Base_Sign": "B" if int(pred_base[i].item()) == 1 else "S",
                    "True_QuoteGap": float(y_gap[i].item()),
                    "Pred_QuoteGap": float(pred_gap[i].item()),
                    "Base_QuoteGap": float(base_gap[i].item()),
                    "LastTradeBaselineUsedFallback": bool(not has_prior_trade[i]),
                })

        n += bs

    summary = {
        "n_samples": int(n),
        "avg_total_loss": total_loss_sum / n,
        "avg_sign_ce": ce_sum / n,
        "avg_gap_huber_log1p": gap_loss_sum / n,

        "model_sign_acc": model_correct / n,
        "baseline_sign_acc": base_correct / n,
        "model_sign_logloss": model_logloss_sum / n,
        "baseline_sign_logloss": base_logloss_sum / n,
        "sign_logloss_gain_vs_baseline": (base_logloss_sum - model_logloss_sum) / n,

        "model_gap_mae": model_mae_sum / n,
        "baseline_gap_mae": base_mae_sum / n,
        "model_gap_rmse": math.sqrt(model_rmse_sum / n),
        "baseline_gap_rmse": math.sqrt(base_rmse_sum / n),

        "baseline_gap_constant_median": float(dataset.gap_median),
        "baseline_sign_fallback_pBuy": float(dataset.sign_prior[1]),
        "frac_no_prior_trade_in_window": n_no_prior_trade / n,
    }

    out_df = pd.DataFrame(rows) if return_df else None
    return summary, out_df


# =========================
# 5. TRAINING
# =========================
def run_training_trade_gap(processed_df: pd.DataFrame):
    assert all(c in processed_df.columns for c in ["Event", "LogTime", "LogSize"]), \
        "processed_df must contain columns: Event, LogTime, LogSize"

    processed_df = processed_df.reset_index(drop=True)

    N = len(processed_df)
    train_end = int(TRAIN_FRAC * N)
    val_end   = int((TRAIN_FRAC + VAL_FRAC) * N)

    train_df = processed_df.iloc[:train_end].reset_index(drop=True)
    val_df   = processed_df.iloc[train_end:val_end].reset_index(drop=True)
    test_df  = processed_df.iloc[val_end:].reset_index(drop=True)

    print(f">>> Split sizes (rows): train={len(train_df)}, val={len(val_df)}, test={len(test_df)}")

    train_dataset = MarketTradeGapDataset(
        train_df,
        seq_len=SEQ_LEN,
        vocab=None,
        add_unk=True,
        banned_history_events=HISTORY_BANNED_EVENTS,
    )
    vocab = train_dataset.vocab

    val_dataset = MarketTradeGapDataset(
        val_df,
        seq_len=SEQ_LEN,
        vocab=vocab,
        add_unk=True,
        banned_history_events=HISTORY_BANNED_EVENTS,
    )
    test_dataset = MarketTradeGapDataset(
        test_df,
        seq_len=SEQ_LEN,
        vocab=vocab,
        add_unk=True,
        banned_history_events=HISTORY_BANNED_EVENTS,
    )

    print(">>> Samples after filtering:")
    print(f"    train={len(train_dataset)}, val={len(val_dataset)}, test={len(test_dataset)}")
    print(f"    train sign prior P(S), P(B) = {train_dataset.sign_prior}")
    print(f"    train quote-gap median      = {train_dataset.gap_median:.3f}")
    print(f"    train quote-gap mean        = {train_dataset.gap_mean:.3f}")

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0),
        drop_last=False,
    )

    model = MarketDecoderOnlyTransformerTradeGap(
        num_event_tokens=train_dataset.num_event_tokens,
        d_model=D_MODEL,
        nhead=NHEAD,
        num_layers=NUM_LAYERS,
        ffn_dim=FFN_DIM,
        dropout=DROPOUT,
        max_seq_len=SEQ_LEN,
    ).to(DEVICE)

    try:
        if hasattr(torch, "compile"):
            model = torch.compile(model)
    except Exception as e:
        print(f">>> torch.compile unavailable or failed (continuing without): {e}")

    optimizer = optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
        betas=(0.9, 0.95),
    )

    ce_loss_fn = nn.CrossEntropyLoss()
    gap_loss_fn = nn.SmoothL1Loss()

    best_val_total = float("inf")
    best_state = None

    print("\n>>> Starting training")
    print("Targets:")
    print("    1) sign of next labeled trade (buy/sell)")
    print("    2) number of quote updates until that trade")
    print(f"History length: {SEQ_LEN}")
    print(f"History banned events: {HISTORY_BANNED_EVENTS}\n")

    for epoch in range(EPOCHS):
        model.train()

        train_total_sum = 0.0
        train_ce_sum = 0.0
        train_gap_sum = 0.0
        train_n = 0

        for x_ev, x_cont, y_sign, y_gap in train_loader:
            bs = x_ev.size(0)

            x_ev = x_ev.to(DEVICE, non_blocking=True)
            x_cont = x_cont.to(DEVICE, non_blocking=True)
            y_sign = y_sign.to(DEVICE, non_blocking=True)
            y_gap = y_gap.to(DEVICE, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)

            with torch.cuda.amp.autocast(enabled=USE_AMP):
                sign_logits, gap_log_pred = model(x_ev, x_cont)
                sign_loss = ce_loss_fn(sign_logits, y_sign)
                gap_loss = gap_loss_fn(gap_log_pred, torch.log1p(y_gap))
                loss = sign_loss + GAP_LOSS_WEIGHT * gap_loss

            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=GRAD_CLIP)
            scaler.step(optimizer)
            scaler.update()

            train_total_sum += float(loss.item()) * bs
            train_ce_sum += float(sign_loss.item()) * bs
            train_gap_sum += float(gap_loss.item()) * bs
            train_n += bs

        train_total = train_total_sum / max(1, train_n)
        train_ce = train_ce_sum / max(1, train_n)
        train_gap = train_gap_sum / max(1, train_n)

        val_summary, _ = evaluate_trade_gap_model(model, val_loader, val_dataset, return_df=False)

        print(
            f"Epoch {epoch+1:02d}/{EPOCHS} | "
            f"Train Total/CE/Gap: {train_total:.4f} / {train_ce:.4f} / {train_gap:.4f} | "
            f"Val Total/CE/Gap: {val_summary['avg_total_loss']:.4f} / "
            f"{val_summary['avg_sign_ce']:.4f} / {val_summary['avg_gap_huber_log1p']:.4f} | "
            f"Val Sign Acc (Model/Base): {val_summary['model_sign_acc']:.4f} / {val_summary['baseline_sign_acc']:.4f} | "
            f"Val Gap MAE (Model/Base): {val_summary['model_gap_mae']:.4f} / {val_summary['baseline_gap_mae']:.4f}"
        )

        if val_summary["avg_total_loss"] < best_val_total:
            best_val_total = val_summary["avg_total_loss"]
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}

    if best_state is not None:
        model.load_state_dict(best_state)

    test_summary, test_pred_df = evaluate_trade_gap_model(
        model, test_loader, test_dataset, return_df=True
    )

    print("\n>>> TEST RESULTS")
    for k, v in test_summary.items():
        if isinstance(v, float):
            print(f"{k}: {v:.6f}")
        else:
            print(f"{k}: {v}")

    return model, test_loader, test_dataset, test_pred_df, test_summary


# =========================
# 6. COLLECT PREDICTIONS
# =========================
@torch.no_grad()
def collect_trade_gap_predictions(model, loader):
    model.eval()
    rows = []

    for x_ev, x_cont, y_sign, y_gap in loader:
        x_ev = x_ev.to(DEVICE, non_blocking=True)
        x_cont = x_cont.to(DEVICE, non_blocking=True)
        y_sign = y_sign.to(DEVICE, non_blocking=True)
        y_gap = y_gap.to(DEVICE, non_blocking=True)

        sign_logits, gap_log_pred = model(x_ev, x_cont)
        sign_probs = torch.softmax(sign_logits, dim=-1)
        pred_sign = torch.argmax(sign_probs, dim=1)
        pred_gap = torch.expm1(gap_log_pred).clamp_min(0.0)

        for i in range(x_ev.size(0)):
            rows.append({
                "True_Sign": "B" if int(y_sign[i].item()) == 1 else "S",
                "Pred_Sign": "B" if int(pred_sign[i].item()) == 1 else "S",
                "Prob_Buy": float(sign_probs[i, 1].item()),
                "True_QuoteGap": float(y_gap[i].item()),
                "Pred_QuoteGap": float(pred_gap[i].item()),
            })

    return pd.DataFrame(rows)


# =========================
# 7. RUN
# =========================
if __name__ == "__main__":
    assert "processed_df" in globals(), "Please define processed_df before running this script."

    trained_model, test_loader, test_dataset, test_pred_df, test_summary = run_training_trade_gap(processed_df)

    pred_df = collect_trade_gap_predictions(trained_model, test_loader)
    print("\n>>> Prediction sample")
    print(pred_df.head())
```
