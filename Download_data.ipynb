```python
import pandas as pd
import pykx as kx

# Connect to the Princeton CME server
q = kx.QConnection(host="hfm.princeton.edu", port=6001)

# Check available tables
print(q("tables[]"))
```

```python
def tqmerge_df(symbol: str):
    dates = ["2024.02.05", "2024.02.06", "2024.02.07", "2024.02.08", "2024.02.09", "2024.02.10", "2024.02.11", "2024.02.12", "2024.02.13", "2024.02.14", "2024.02.15", "2024.02.16"]
    dfs = []
    
    for d in dates:
        res = q(f"tqmergeT[{d}; `{symbol}; 00:00:00; 23:59:59]")
        dfs.append(res.pd())
    
    merged_all = pd.concat(dfs, ignore_index=True)
    print(len(merged_all))
    
    # merged_all.to_parquet(f"{symbol}_merged_quotes_trades.parquet")
    
```

```python
tqmerge_df('CLJ4')
tqmerge_df('HOJ4')
tqmerge_df('SR3J4')
```
